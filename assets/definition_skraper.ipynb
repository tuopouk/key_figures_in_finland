```python
import pandas as pd
import geopandas as gpd
import requests
import numpy as np
import orjson
from bs4 import BeautifulSoup
```

```python
for lang in ['en','fi','sv']:

    url = f"https://pxdata.stat.fi/PxWeb/pxweb/{lang}/Kuntien_avainluvut/Kuntien_avainluvut__uusin/kuntien_avainluvut_viimeisin.px/"

    definitions = []

    soup = BeautifulSoup(requests.get(url).content)
    for h4 in soup.findAll('h4'):
        

        definitions.append({'key_figure':h4.text.split(',')[0],'definition':h4.next.next.text})

    pd.DataFrame(definitions).set_index('key_figure').to_csv(f'definitions_{lang}.csv', encoding = 'utf-8')
```

```python
from gtts import gTTS
from tqdm import tqdm
import re
import time
for lang in ['sv','en','fi']:
    
    definitions = pd.read_csv(f'definitions_{lang}.csv').set_index('key_figure')
    
    for kw in tqdm(definitions.index):
        
        try:
            my_speech = gTTS(text=definitions.loc[kw].definition, lang=lang, slow=False)
        except:
            time.sleep(5)
            my_speech = gTTS(text=definitions.loc[kw].definition, lang=lang, slow=False)
        kw = re.sub(r'\W+','',kw)

        filename = f"recordings/{lang}/{kw}.mp3"
        my_speech.save(filename)
    
```

```python
def get_region_names():
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/83.0.4103.97 Safari/537.36',
 'Content-Type': 'application/json'}
    df = None
    for lang in ['fi','sv','en']:
        url = f"https://pxdata.stat.fi:443/PxWeb/api/v1/{lang}/Kuntien_avainluvut/2021/kuntien_avainluvut_2021_viimeisin.px"
        json = requests.get(url, headers=headers).json()
        dff = pd.DataFrame([{'id':i, {'fi':'nimi','sv':'namn','en':'name'}[lang]:name} for i,name in zip(json['variables'][0]['values'],json['variables'][0]['valueTexts'])])
        if df is None:
            df = dff
        else:
            df = pd.merge(left = df, right = dff, on = 'id', how = 'inner')
    df = df.astype({'id':'category','nimi':'category', 'namn':'category', 'name':'category'})
    return df.set_index('id')
reg_names = get_region_names()

reg_names
```
